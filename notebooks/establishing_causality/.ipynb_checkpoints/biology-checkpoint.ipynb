```python
projection_diff_list = []

for i in range(len(harmless_prompts)):

    harmless_prompt = get_chat_template(harmless_prompts[i], tokenizer)
    harmful_prompt = get_chat_template(harmful_prompts[i], tokenizer)
    
    harmless_projection = get_refusal_projection(gemma2, gemma2_sae, harmless_prompt, refusal_direction)
    harmful_projection = get_refusal_projection(gemma2, gemma2_sae, harmful_prompt, refusal_direction)
    
    projection_diff = harmful_projection - harmless_projection
    projection_diff_list.append(projection_diff)
```

```python
layer = 5
# prompt = harmful_prompts[0]
prompt = harmless_prompts[0]
prompt = get_chat_template(prompt, tokenizer)

# index = gemma2.to_str_tokens(prompt).index(' killing')
hugging_index = gemma2.to_str_tokens(prompt, prepend_bos = False).index(' hugging')
```

```python
_, cache = gemma2.run_with_cache_with_saes(
    prompt,
    saes=[gemma2_sae],
    stop_at_layer=gemma2_sae.cfg.hook_layer + 1,
    prepend_bos = False
)
sae_acts_post = cache[f"{gemma2_sae.cfg.hook_name}.hook_sae_acts_post"]

t.set_grad_enabled(True)
gradient = get_gradient(gemma2, prompt, layer, refusal_direction, prepend_bos = False)
```

```python
count = 0

for i in range(sae_acts_post.shape[1]):
    index = i

    print(gemma2.to_str_tokens(prompt, prepend_bos = False)[index])

    gradient_idx = gradient[:, index, :]
    norm = t.norm(gradient_idx, p=2)
    gradient_idx_unit = gradient_idx/norm

    sae_dec_projection = einops.einsum(gemma2_sae.W_dec.detach(), gradient_idx_unit.squeeze(), "d_sae d_in, d_in -> d_sae")

    print(t.topk(sae_dec_projection, 10))

    if 12072 in t.topk(sae_dec_projection, 10).indices:
        count += 1
```

```python
def repeat_elements(input_list, n):
    result = []
    for element in input_list:
        result.extend([element] * n)
    return result
```

## Can we find other latents?

```python
top_10_list_all = []

for i in range(sae_acts_post.shape[1]):
    index = i
    gradient_idx = gradient[:, index, :]
    norm = t.norm(gradient_idx, p=2)
    gradient_idx_unit = gradient_idx/norm

    sae_dec_projection = einops.einsum(gemma2_sae.W_dec.detach(), gradient_idx_unit.squeeze(), "d_sae d_in, d_in -> d_sae")

    top_10_list = t.topk(sae_dec_projection, 10).indices
    top_10_list_all.extend(top_10_list)

```

```python
top_10_list_all = [x.item() for x in top_10_list_all]
```

```python
from collections import Counter

# Count occurrences
counts = Counter(top_10_list_all)

# Sort by frequency in descending order
sorted_counts = counts.most_common()

# Print the result
print(sorted_counts)
```
